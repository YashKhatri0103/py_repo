```python
my_list = [10, 20, 30, 40]
print("original list",my_list)
```

```python
my_list.append(50)
print(my_list)
```

```python
my_list.insert(1, 15)
print(my_list)
```

```python
my_list.remove(40)
print(my_list)
```

```python
my_list.pop()
print(my_list)
```

```python
my_list.sort()
print(my_list)
```

```python
print("List after operations:", my_list)
```

```python
my_tuple = (10, 20, 30, 20, 40)
print(my_tuple) 
```

```python
print("Indexing [2]:", my_tuple[2])
```

```python
print("Slicing [1:4]:", my_tuple [1:4])
```

```python
print("Count of 20:", my_tuple.count(20))
```

```python
print("Index of 30:", my_tuple.index(30))
```

```python
new_tuple = my_tuple + (50, 60)
print("After concatenation:", new_tuple)
```
