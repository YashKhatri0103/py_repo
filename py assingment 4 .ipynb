```python
# Matrix addition using NumPy

import numpy as np

matrix1 = np.array([[1, 2],
                    [3, 4]])

matrix2 = np.array([[5, 6],
                    [7, 8]])

result = matrix1 + matrix2

print("\nAddition of matrices using NumPy:")
print(result)
```

```python

# Matrix subtraction using NumPy

import numpy as np

matrix1 = np.array([[1, 2],
                    [3, 4]])

matrix2 = np.array([[5, 6],
                    [7, 8]])

result = matrix1 - matrix2

print("Subtraction of matrices using NumPy:")
print(result)
```

```python

# Matrix multiplication using NumPy

import numpy as np

matrix1 = np.array([[1, 2],
                    [3, 4]])

matrix2 = np.array([[5, 6],
                    [7, 8]])

result = np.dot(matrix1, matrix2)

print("Multiplication of matrices using NumPy:")
print(result)
```

```python

# Determinant of a matrix using NumPy

import numpy as np

matrix = np.array([[1, 2],
                   [3, 4]])

result = np.linalg.det(matrix)

print("Determinant of the matrix:")
print(result)
```

```python

# Inverse of a matrix using NumPy

import numpy as np

matrix = np.array([[1, 2],
                   [3, 4]])

result = np.linalg.inv(matrix)

print("Inverse of the matrix:")
print(result)
```

```python

# Adjoint (Adjugate) of a matrix using NumPy

import numpy as np

def adjoint(matrix):
    n = matrix.shape[0]
    adj = np.zeros((n, n), dtype=float)

    if n == 1:
        adj[0, 0] = 1
        return adj

    for i in range(n):
        for j in range(n):
            minor = np.delete(np.delete(matrix, i, axis=0), j, axis=1)
            cofactor = ((-1) ** (i + j)) * np.linalg.det(minor)
            adj[j, i] = cofactor

    return adj

matrix = np.array([[1, 2],
                   [3, 4]])

result = adjoint(matrix)

print("Adjoint of the matrix:")
print(result)
```

```python

# Transpose of a matrix using NumPy

import numpy as np

matrix = np.array([[1, 2],
                   [3, 4]])

result = np.transpose(matrix)

print("Transpose of the matrix:")
print(result)

```
